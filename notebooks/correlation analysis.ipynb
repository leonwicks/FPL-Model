## Build a Simple Model

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

cols_to_drop = [
    'total_points',
    'bps',
    'bonus',
    'goals_scored',
    'minutes',
    'starts',
    'clean_sheets',
    'assists',
    'selected',
    'saves',
    'yellow_cards',
    'penalties_saved',
    'red_cards',
    'team_a_score',
    'own_goals',
    'round',
    'team_h_score',
    'id_player',
    'modified',
    'rolling_mean_modified_3',
    'rolling_mean_modified_5',
    'rolling_mean_modified_7',
    'web_name',
    'influence',
    'creativity',
    'threat',
    'ict_index',
    'expected_goals',
    'expected_assists',
    'expected_goal_involvements',
    'expected_goals_conceded',
]

df_proc = df_rm_pos.fillna(0)

df_latest = df_proc[df_proc['round'] == df_proc['round'].max()]
df_proc = df_proc[~df_proc.index.isin(df_latest.index)]

df_proc_test = df_proc.sample(frac=0.2, random_state=42)
df_proc_train = df_proc[~df_proc.index.isin(df_proc_test.index)]

X_train = df_proc_train.drop(columns=cols_to_drop).values
y_train = df_proc_train['total_points'].values

X_test = df_proc_test.drop(columns=cols_to_drop).values
y_test = df_proc_test['total_points'].values

lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, mean_absolute_percentage_error

y_pred = lr.predict(X_test)

rmse = root_mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
mape = mean_absolute_percentage_error(y_test, y_pred)

print(
    f"RMSE: {rmse}\n"
    f"MAE: {mae}\n"
    f"MAPE: {mape}"
)
```

```python
df_proc_test['total_points_pred'] = y_pred

cols_to_display = [
    'web_name',
    'opponent_team',
    'total_points',
    'total_points_pred'
]

df_temp = df_proc_test[cols_to_display]
df_temp = df_temp.merge(
    df[['team', 'team_name']],
    how='inner',
    left_on='opponent_team',
    right_on='team'
).drop_duplicates().drop(columns=['opponent_team','team'])

df_temp['diff'] = df_temp['total_points_pred'] - df_temp['total_points']
df_temp.sort_values(by='diff', ascending=False).head(20)
```

```python
X_latest = df_latest.drop(columns=cols_to_drop).values
y_latest = df_latest['total_points'].values

y_latest = lr.predict(X_latest)
df_latest['total_points_pred'] = y_latest

df_latest = df_latest.merge(
    df[['id_player','player_position']],
    how='inner',
    left_on='id_player',
    right_on='id_player'
).drop_duplicates()

cols_to_display.append('player_position')

df_latest.sort_values(by='total_points_pred', ascending=False)[cols_to_display].head(20)
```

## Build Seperate Models for Each Position

```python
df_latest = df_proc[df_proc['round'] == df_proc['round'].max()]
df_proc = df_proc[~df_proc.index.isin(df_latest.index)]

df_proc_test = df_proc.sample(frac=0.2, random_state=42)
df_proc_train = df_proc[~df_proc.index.isin(df_proc_test.index)]

gkp_ids = df[df['player_position'] == 'GKP']['id_player'].unique()
def_ids = df[df['player_position'] == 'DEF']['id_player'].unique()
mid_ids = df[df['player_position'] == 'MID']['id_player'].unique()
fwd_ids = df[df['player_position'] == 'FWD']['id_player'].unique()

df_proc_train_gkp = df_proc_train[df_proc_train['id_player'].isin(gkp_ids)]
df_proc_train_def = df_proc_train[df_proc_train['id_player'].isin(def_ids)]
df_proc_train_mid = df_proc_train[df_proc_train['id_player'].isin(mid_ids)]
df_proc_train_fwd = df_proc_train[df_proc_train['id_player'].isin(fwd_ids)]

df_proc_test_gkp = df_proc_test[df_proc_test['id_player'].isin(gkp_ids)]
df_proc_test_def = df_proc_test[df_proc_test['id_player'].isin(def_ids)]
df_proc_test_mid = df_proc_test[df_proc_test['id_player'].isin(mid_ids)]
df_proc_test_fwd = df_proc_test[df_proc_test['id_player'].isin(fwd_ids)]

X_train_gkp = df_proc_train_gkp.drop(columns=cols_to_drop).values
X_train_def = df_proc_train_def.drop(columns=cols_to_drop).values
X_train_mid = df_proc_train_mid.drop(columns=cols_to_drop).values
X_train_fwd = df_proc_train_fwd.drop(columns=cols_to_drop).values

y_train_gkp = df_proc_train_gkp['total_points'].values
y_train_def = df_proc_train_def['total_points'].values
y_train_mid = df_proc_train_mid['total_points'].values
y_train_fwd = df_proc_train_fwd['total_points'].values

X_test_gkp = df_proc_test_gkp.drop(columns=cols_to_drop).values
X_test_def = df_proc_test_def.drop(columns=cols_to_drop).values
X_test_mid = df_proc_test_mid.drop(columns=cols_to_drop).values
X_test_fwd = df_proc_test_fwd.drop(columns=cols_to_drop).values

y_test_gkp = df_proc_test_gkp['total_points'].values
y_test_def = df_proc_test_def['total_points'].values
y_test_mid = df_proc_test_mid['total_points'].values
y_test_fwd = df_proc_test_fwd['total_points'].values

lr_gkp = LinearRegression()
lr_def = LinearRegression()
lr_mid = LinearRegression()
lr_fwd = LinearRegression()

lr_gkp.fit(X_train_gkp, y_train_gkp)
lr_def.fit(X_train_def, y_train_def)
lr_mid.fit(X_train_mid, y_train_mid)
lr_fwd.fit(X_train_fwd, y_train_fwd)

y_pred_gkp = lr_gkp.predict(X_test_gkp)
y_pred_def = lr_def.predict(X_test_def)
y_pred_mid = lr_mid.predict(X_test_mid)
y_pred_fwd = lr_fwd.predict(X_test_fwd)

print(
    "For the Goalkeeper Model:\n"
    f"RMSE: {root_mean_squared_error(y_test_gkp, y_pred_gkp)}\n"
    f"MAE: {mean_absolute_error(y_test_gkp, y_pred_gkp)}\n"
    f"MAPE: {mean_absolute_percentage_error(y_test_gkp, y_pred_gkp)}\n\n"

    "For the Defender Model:\n"
    f"RMSE: {root_mean_squared_error(y_test_def, y_pred_def)}\n"
    f"MAE: {mean_absolute_error(y_test_def, y_pred_def)}\n"
    f"MAPE: {mean_absolute_percentage_error(y_test_def, y_pred_def)}\n\n"

    "For the Midfielder Model:\n"
    f"RMSE: {root_mean_squared_error(y_test_mid, y_pred_mid)}\n"
    f"MAE: {mean_absolute_error(y_test_mid, y_pred_mid)}\n"
    f"MAPE: {mean_absolute_percentage_error(y_test_mid, y_pred_mid)}\n\n"

    "For the Forward Model:\n"
    f"RMSE: {root_mean_squared_error(y_test_fwd, y_pred_fwd)}\n"
    f"MAE: {mean_absolute_error(y_test_fwd, y_pred_fwd)}\n"
    f"MAPE: {mean_absolute_percentage_error(y_test_fwd, y_pred_fwd)}\n\n"
)
```

```python
df_latest = df_latest.merge(
    df[['id_player','player_position']],
    how='inner',
    left_on='id_player',
    right_on='id_player',
).drop_duplicates()

df_latest_gkp = df_latest[df_latest['player_position'] == 'GKP']
df_latest_def = df_latest[df_latest['player_position'] == 'DEF']
df_latest_mid = df_latest[df_latest['player_position'] == 'MID']
df_latest_fwd = df_latest[df_latest['player_position'] == 'FWD']

df_latest_gkp.drop('player_position', axis=1, inplace=True)
df_latest_def.drop('player_position', axis=1, inplace=True)
df_latest_mid.drop('player_position', axis=1, inplace=True)
df_latest_fwd.drop('player_position', axis=1, inplace=True)

X_latest_gkp = df_latest_gkp.drop(columns=cols_to_drop).values
X_latest_def = df_latest_def.drop(columns=cols_to_drop).values
X_latest_mid = df_latest_mid.drop(columns=cols_to_drop).values
X_latest_fwd = df_latest_fwd.drop(columns=cols_to_drop).values

y_latest_gkp = df_latest_gkp['total_points'].values
y_latest_def = df_latest_def['total_points'].values
y_latest_mid = df_latest_mid['total_points'].values
y_latest_fwd = df_latest_fwd['total_points'].values

y_pred_latest_gkp = lr_gkp.predict(X_latest_gkp)
y_pred_latest_def = lr_def.predict(X_latest_def)
y_pred_latest_mid = lr_mid.predict(X_latest_mid)
y_pred_latest_fwd = lr_fwd.predict(X_latest_fwd)

df_latest_gkp['total_points_pred'] = y_pred_latest_gkp
df_latest_def['total_points_pred'] = y_pred_latest_def
df_latest_mid['total_points_pred'] = y_pred_latest_mid
df_latest_fwd['total_points_pred'] = y_pred_latest_fwd

df_latest_pred = pd.concat([df_latest_gkp, df_latest_def, df_latest_mid, df_latest_fwd])

df_latest_pred = df_latest_pred.merge(
    df[['id_player', 'team_name', 'now_cost', 'chance_of_playing_next_round', 'player_position']],
    how='inner',
    left_on='id_player',
    right_on='id_player',
).drop_duplicates()

squad = select_fpl_squad(df_latest_pred, metric='total_points_pred')
```
